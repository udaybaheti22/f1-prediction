### Adding kaggle_driverid to the github season driver dataset

```python
def normalize_name_series(s):
    return (
        s.str.lower()
         .str.strip()
         .str.normalize('NFKD')
         .str.encode('ascii', errors='ignore')
         .str.decode('utf-8')
    )

def add_kaggle_driver_id(df, kaggle_driver):
    df = df.copy()

    df['driver_norm'] = normalize_name_series(df['Driver'])

    kd = kaggle_driver.copy()
    kd['driver_norm'] = normalize_name_series(
        kd['forename'] + ' ' + kd['surname']
    )

    df = df.merge(
        kd[['driverId', 'driver_norm']],
        on='driver_norm',
        how='left'
    )

    df.rename(columns={'driverId': 'kaggle_driver_id'}, inplace=True)
    df.drop(columns=['driver_norm'], inplace=True)

    return df
```

### Only keeping kaggle_driverId , driver name and their team in the driver dataset according to the season

```python
def process_season_drivers(df, year, kaggle_driver):
    df = df.copy()

    # Keep only required columns
    df = df[['Driver', 'Team']]

    # Add kaggle_driver_id
    df = add_kaggle_driver_id(df, kaggle_driver)

    # Reorder columns
    df = df[['kaggle_driver_id', 'Driver', 'Team']]

    # Save to root folder
    filename = f"f1{year}season_drivers_processed.csv"
    df.to_csv(filename, index=False)

    print(f"Saved: {filename}")

    return df
```

## Assigning teams

```python
def standardize_team(df, year):
    df = df.copy()

    # Add year column if missing
    if "year" not in df.columns:
        df["year"] = year

    # Drop existing Team if present
    if "Team" in df.columns:
        df = df.drop(columns=["Team"])

    # Merge standardized team
    df = df.merge(
        team_master,
        on=["kaggle_driver_id", "year"],
        how="left"
    )

    # Validation check
    missing = df["Team"].isna().sum()
    if missing > 0:
        print(f"Warning: {missing} team values missing for year {year}")

    return df
```

### Splitting the dataset by using unique raceId from year 2019,2020 and 2021

```python
def validate_and_split_by_year(df, race_ids, year):
    df_year = df[df["raceId"].isin(race_ids)].copy()

    extracted_ids = set(df_year["raceId"].unique())

    missing = race_ids - extracted_ids
    extra = extracted_ids - race_ids

    if missing:
        raise ValueError(f"[{year}] Missing raceIds: {missing}")

    if extra:
        raise ValueError(f"[{year}] Extra raceIds detected: {extra}")

    print(f"[{year}] Validation passed. Races: {len(extracted_ids)}")

    return df_year
```

### Converting quali times with minor changes

```python
def quali_time_to_sec(val):
    if pd.isna(val):
        return np.nan

    val = str(val).strip()
    parts = val.split(':')

    try:
        if len(parts) == 2:
            # mm:ss.xxx
            m, s = parts
            return int(m) * 60 + float(s)
        elif len(parts) == 3:
            # hh:mm:ss.xxx (rare but exists)
            h, m, s = parts
            return int(h) * 3600 + int(m) * 60 + float(s)
    except:
        return np.nan

    return np.nan
def process_quali(df):
    df = df.copy()
    # in kaggle dataset in empty fields of q2 and q3, its not NaN it is /N so we replace by NA
    df[['q1', 'q2', 'q3']] = df[['q1', 'q2', 'q3']].replace(r'\\N', pd.NA, regex=True)
    # participation flags
    df['participated_q2'] = df['q2'].notna().astype(int)
    df['participated_q3'] = df['q3'].notna().astype(int)

    # parse times
    df['q1_time_sec'] = df['q1'].apply(quali_time_to_sec)
    df['q2_time_sec'] = df['q2'].apply(quali_time_to_sec)
    df['q3_time_sec'] = df['q3'].apply(quali_time_to_sec)

    # keep final columns
    df = df[
        [
            'raceId',
            #'Track',
            'Team',
            'kaggle_driver_id',
            'position',
            'participated_q2',
            'participated_q3',
            'q1_time_sec',
            'q2_time_sec',
            'q3_time_sec',
        ]
    ]

    return df
```

```python
kq_2019 = process_quali(kq_2019)
kq_2020 = process_quali(kq_2020)
kq_2021 = process_quali(kq_2021)
```

### The following datasets miss Track column

```python
def check_track_column(datasets):
    missing_track = []
    present_track = []

    for name, df in datasets.items():
        # Case-insensitive column check
        cols_lower = [col.lower() for col in df.columns]

        if "track" in cols_lower:
            present_track.append(name)
        else:
            missing_track.append(name)

    print("✅ Datasets WITH Track column:")
    for name in present_track:
        print("  -", name)

    print("\n❌ Datasets MISSING Track column:")
    for name in missing_track:
        print("  -", name)

    return present_track, missing_track

present, missing = check_track_column(datasets)
```

### We add Track by mapping the raceId

```python
def add_track_column(df, mapping):
    df = df.copy()
    
    # Drop existing Track if present (avoid duplication)
    if "Track" in df.columns:
        df = df.drop(columns=["Track"])
    
    df = df.merge(mapping, on="raceId", how="left")
    
    # Safety check
    missing = df["Track"].isna().sum()
    if missing > 0:
        raise ValueError(f"Track missing for {missing} rows.")
    
    return df
```

```python
kaggle_sprint = kaggle_sprint.sort_values(['raceId']).reset_index(drop=True)
```

```python
sorted(kaggle_sprint['raceId'].unique())
```

```python
def assign_sprint_race_ids(sprint_dfs, kaggle_sprint):

    # 1. Get ordered sprint raceIds
    sprint_race_ids = sorted(kaggle_sprint['raceId'].unique())
    
    race_idx = 0
    out = []

    for df in sprint_dfs:
        df = df.copy()
        race_ids = []
        i = 0
        n = len(df)

        while i < n:
            if race_idx >= len(sprint_race_ids):
                raise IndexError("Not enough sprint raceIds available")

            race_id = sprint_race_ids[race_idx]

            size = 19 if race_id == 1101 else 20

            if i + size > n:
                raise ValueError("Block mismatch — sprint dataset misaligned")

            race_ids.extend([race_id] * size)

            i += size
            race_idx += 1

        if len(race_ids) != n:
            raise ValueError("Assignment length mismatch")

        df['raceId'] = race_ids
        out.append(df)

    print("✅ Sprint raceId assignment completed cleanly")
    return out
```

```python
sprint_race_res_2022, sprint_race_res_2023, sprint_race_res_2024 = assign_sprint_race_ids(
    [sprint_race_res_2022, sprint_race_res_2023, sprint_race_res_2024],
    kaggle_sprint
)
```
