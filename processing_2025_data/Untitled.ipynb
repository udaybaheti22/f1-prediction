```python
def assign_future_race_ids(df, starting_race_id):
    df = df.copy()

    race_ids = []
    current_race_id = starting_race_id

    tracks = df['Track'].values
    n = len(df)

    race_ids.append(current_race_id)

    for i in range(1, n):
        if tracks[i] != tracks[i - 1]:
            current_race_id += 1

        race_ids.append(current_race_id)

    df['raceId'] = race_ids
    return df
```

```python
"""
F1 2025 Season Weather Enrichment (v2) — using real local race start times
----------------------------------------------------------------------------
Adds 'rain' and 'sunny' columns to the 2025 calendar (with raceId), converting
each race's LOCAL start time to UTC (via each circuit's IANA timezone, so DST
is handled automatically) before querying the Open-Meteo archive API.

Usage in notebook:
    df = pd.read_csv("Formula1_2025Season_Calendar.csv")   # has raceId already
    weather_df = build_weather_df(df)          # -> raceId, rain, sunny
    race_results = add_weather(race_results)   # your merge function, unchanged
"""

import requests
import pandas as pd
import time
from datetime import datetime
from zoneinfo import ZoneInfo

# ---------------------------------------------------------------------------
# Per-circuit: (lat, lon, IANA timezone, official local start time "HH:MM")
# ---------------------------------------------------------------------------
RACE_INFO_2025 = {
    "Albert Park Circuit":                         (-37.8497,  144.9680, "Australia/Melbourne", "15:00"),
    "Shanghai International Circuit":              ( 31.3389,  121.2198, "Asia/Shanghai",        "15:00"),
    "Suzuka Circuit":                               ( 34.8431,  136.5407, "Asia/Tokyo",           "14:00"),
    "Bahrain International Circuit":               ( 26.0325,   50.5106, "Asia/Bahrain",         "18:00"),
    "Jeddah Corniche Circuit":                      ( 21.6319,   39.1044, "Asia/Riyadh",          "20:00"),
    "Miami International Autodrome":                ( 25.9581,  -80.2389, "America/New_York",     "16:00"),
    "Autodromo Internazionale Enzo e Dino Ferrari": ( 44.3439,   11.7167, "Europe/Rome",          "15:00"),
    "Circuit de Monaco":                            ( 43.7347,    7.4205, "Europe/Monaco",        "15:00"),
    "Circuit de Barcelona-Catalunya":               ( 41.5700,    2.2611, "Europe/Madrid",        "15:00"),
    "Circuit Gilles-Villeneuve":                    ( 45.5000,  -73.5228, "America/Toronto",      "14:00"),
    "Red Bull Ring":                                ( 47.2197,   14.7647, "Europe/Vienna",        "15:15"),
    "Silverstone Circuit":                          ( 52.0786,   -1.0169, "Europe/London",        "15:00"),
    "Circuit de Spa-Francorchamps":                 ( 50.4372,    5.9714, "Europe/Brussels",      "15:00"),
    "Hungaroring":                                  ( 47.5830,   19.2526, "Europe/Budapest",      "15:00"),
    "Circuit Zandvoort":                            ( 52.3888,    4.5409, "Europe/Amsterdam",     "15:00"),
    "Autodromo Nazionale Monza":                    ( 45.6156,    9.2811, "Europe/Rome",          "15:00"),
    "Baku City Circuit":                            ( 40.3725,   49.8533, "Asia/Baku",            "15:00"),
    "Marina Bay Street Circuit":                    (  1.2914,  103.8640, "Asia/Singapore",       "20:00"),
    "Circuit of The Americas":                      ( 30.1328,  -97.6411, "America/Chicago",      "14:00"),
    "Autodromo Hermanos Rodriguez":                 ( 19.4042,  -99.0907, "America/Mexico_City",  "14:00"),
    "Aurodromo Jose Carlos Pace":                   (-23.7036,  -46.6997, "America/Sao_Paulo",    "14:00"),  # typo variant in source
    "Autodromo Jose Carlos Pace":                   (-23.7036,  -46.6997, "America/Sao_Paulo",    "14:00"),  # correct spelling
    "Las Vegas Strip Circuit":                      ( 36.1147, -115.1728, "America/Los_Angeles",  "20:00"),
    "Lusail International Circuit":                 ( 25.4900,   51.4542, "Asia/Qatar",           "19:00"),
    "Yas Marina Circuit":                           ( 24.4672,   54.6031, "Asia/Dubai",           "17:00"),
}

RAIN_THRESHOLD_MM   = 0.1
SUNNY_CLOUD_MAX_PCT = 65
RACE_DURATION_HRS   = 2


def local_to_utc(date_val, local_time_str: str, tz_name: str):
    """Combine a date + local HH:MM start time in tz_name, return (utc_date_str, utc_time_str)."""
    date_obj = pd.to_datetime(date_val).date()
    hour, minute = map(int, local_time_str.split(":"))
    local_dt = datetime(date_obj.year, date_obj.month, date_obj.day, hour, minute, tzinfo=ZoneInfo(tz_name))
    utc_dt = local_dt.astimezone(ZoneInfo("UTC"))
    return utc_dt.strftime("%Y-%m-%d"), utc_dt.strftime("%H:%M:%S")


def fetch_weather(lat: float, lon: float, date: str, start_time: str, gp_name: str = "") -> dict:
    """Calls Open-Meteo archive API and returns rain/sunny flags. date/start_time must be UTC."""
    url = "https://archive-api.open-meteo.com/v1/archive"
    params = {
        "latitude":   lat,
        "longitude":  lon,
        "start_date": date,
        "end_date":   date,
        "hourly":     "precipitation,cloudcover",
        "timezone":   "UTC",
    }
    resp = requests.get(url, params=params, timeout=15)
    resp.raise_for_status()
    data = resp.json()

    hours      = data["hourly"]["time"]
    precip     = data["hourly"]["precipitation"]
    cloudcover = data["hourly"]["cloudcover"]

    start_hour = int(start_time.split(":")[0])
    end_hour   = min(start_hour + RACE_DURATION_HRS, 23)

    race_precip = [precip[i] for i, t in enumerate(hours) if start_hour <= int(t[11:13]) <= end_hour]
    race_cloud  = [cloudcover[i] for i, t in enumerate(hours) if start_hour <= int(t[11:13]) <= end_hour]

    if any(p is None for p in race_precip) or any(c is None for c in race_cloud):
        print(f"    [WARN] {gp_name} {date} — nulls in race window")

    rain  = int(any(p is not None and p > RAIN_THRESHOLD_MM for p in race_precip))
    sunny = int(any(c is not None and c < SUNNY_CLOUD_MAX_PCT for c in race_cloud))

    if rain == 0 and sunny == 0:
        sunny = 1

    return {"rain": rain, "sunny": sunny}


def build_weather_df(calendar_df: pd.DataFrame) -> pd.DataFrame:
    """
    Builds a small (raceId, rain, sunny) dataframe from the 2025 calendar,
    using each race's real local start time converted to UTC.

    Expects columns: 'raceId', 'Race Date', 'Circuit Name', 'GP Name'.
    """
    rows = []

    for _, row in calendar_df.iterrows():
        race_id      = row["raceId"]
        gp_name      = row["GP Name"]
        circuit_name = row["Circuit Name"]
        race_date    = row["Race Date"]

        if circuit_name not in RACE_INFO_2025:
            print(f"  [SKIP] No info for circuit: {circuit_name}")
            rows.append({"raceId": race_id, "rain": None, "sunny": None})
            continue

        lat, lon, tz_name, local_time = RACE_INFO_2025[circuit_name]
        utc_date, utc_time = local_to_utc(race_date, local_time, tz_name)

        try:
            result = fetch_weather(lat, lon, utc_date, utc_time, gp_name)
            rows.append({"raceId": race_id, "rain": result["rain"], "sunny": result["sunny"]})
            print(f"  [OK] {gp_name:55s} | local {local_time} {tz_name:22s} -> UTC {utc_date} {utc_time} | rain={result['rain']} sunny={result['sunny']}")
        except Exception as e:
            print(f"  [ERR] {gp_name} | {e}")
            rows.append({"raceId": race_id, "rain": None, "sunny": None})

        time.sleep(0.3)

    return pd.DataFrame(rows)


# ---------------------------------------------------------------------------
# Example usage in a notebook cell:
# ---------------------------------------------------------------------------
# calendar_df = pd.read_csv("Formula1_2025Season_Calendar.csv")   # must include raceId
# weather_df = build_weather_df(calendar_df)                       # raceId, rain, sunny
#
# def add_weather(data):
#     return data.merge(weather_df, on="raceId", how="left")
#
# race_results = add_weather(race_results)
```
