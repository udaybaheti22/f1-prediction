## Validating Columns

```python
def validate_columns(datasets, name):
    print(f"\nChecking column consistency for: {name}")
    
    base_cols = datasets[0].columns.tolist()
    
    for i, df in enumerate(datasets):
        if df.columns.tolist() != base_cols:
            print(f"❌ Mismatch in dataset index {i}")
            print("Expected:", base_cols)
            print("Found   :", df.columns.tolist())
            return
    print("✅ All datasets have identical columns and order.")
```

```python
validate_columns(main_race_results, "Main Race Results")
validate_columns(main_race_quali, "Main Race Qualifying")
validate_columns(sprint_results, "Sprint Race Results")
validate_columns(sprint_quali, "Sprint Qualifying")
```

### Making a fixed schema

```python
def clean_main_quali(df):
    df = df.copy()

    # Drop Unnamed columns
    df = df.loc[:, ~df.columns.str.contains('^Unnamed')]

    # Standardize column names
    df.columns = df.columns.str.strip()

    if "Position" in df.columns:
        df.rename(columns={"Position": "position"}, inplace=True)

    # Drop year if exists
    if "year" in df.columns:
        df.drop(columns=["year"], inplace=True)

    # Reorder
    df = df[main_race_quali_cols]

    return df
```

```python
main_race_quali_cols = [
    'raceId', 'Team', 'kaggle_driver_id', 'position',
    'participated_q2', 'participated_q3',
    'q1_time_sec', 'q2_time_sec', 'q3_time_sec', 'Track'
]
```

```python
sprint_results_cols = [
    'Position', 'Starting Grid', 'kaggle_driver_id',
    'raceId', 'finished', 'dnf', 'laps_down',
    'time_gap_sec', 'year', 'Team', 'Track'
]
```

```python
sprint_quali_cols = [
    'raceId', 'Track', 'kaggle_driver_id', 'Position',
    'participated_q2', 'participated_q3',
    'q1_time_sec', 'q2_time_sec', 'q3_time_sec',
    'year', 'Team'
]
```

```python
main_race_quali_2019 = clean_main_quali(main_race_quali_2019)
main_race_quali_2020 = clean_main_quali(main_race_quali_2020)
main_race_quali_2021 = clean_main_quali(main_race_quali_2021)
main_race_quali_2022 = clean_main_quali(main_race_quali_2022)
main_race_quali_2023 = clean_main_quali(main_race_quali_2023)
main_race_quali_2024 = clean_main_quali(main_race_quali_2024)
```

```python
def clean_sprint_results(df):
    df = df.copy()
    df = df[sprint_results_cols]
    return df
```

```python
def clean_sprint_quali(df):
    df = df.copy()
    df = df.loc[:, ~df.columns.str.contains('^Unnamed')]
    df = df[sprint_quali_cols]
    return df
```

```python
sprint_race_res_2021 = clean_sprint_results(sprint_race_res_2021)
sprint_race_res_2022 = clean_sprint_results(sprint_race_res_2022)
sprint_race_res_2023 = clean_sprint_results(sprint_race_res_2023)
sprint_race_res_2024 = clean_sprint_results(sprint_race_res_2024)
```

```python
sprint_quali_2023 = clean_sprint_quali(sprint_quali_2023)
sprint_quali_2024 = clean_sprint_quali(sprint_quali_2024)
```

```python
validate_columns(main_race_results, "Main Race Results")
validate_columns(main_race_quali, "Main Race Qualifying")
validate_columns(sprint_results, "Sprint Race Results")
validate_columns(sprint_quali, "Sprint Qualifying")
```

## Checking for missing values

```python
def check_missing_values(datasets, name):
    print(f"\n{name}")
    
    for i, df in enumerate(datasets):
        print(f"\nDataset index {i}")
        
        for col in ['raceId', 'kaggle_driver_id', 'Team', 'Track']:
            if col in df.columns:
                missing = df[col].isna().sum()
                print(f"{col}: {missing}")
```

```python
check_missing_values(main_race_results, "Main Race Results")
check_missing_values(main_race_quali, "Main Race Qualifying")
check_missing_values(sprint_results, "Sprint Race Results")
check_missing_values(sprint_quali, "Sprint Qualifying")
```

## Checking for Duplicate Driver per Race

```python
def check_duplicates(datasets, name):
    print(f"\n{name}")
    
    for i, df in enumerate(datasets):
        if "raceId" in df.columns:
            dup = df.duplicated(subset=["raceId", "kaggle_driver_id"]).sum()
            print(f"Dataset {i} duplicates:", dup)
```

```python
check_duplicates(main_race_results, "Main Race Results")
check_duplicates(main_race_quali, "Main Race Qualifying")
check_duplicates(sprint_results, "Sprint Race Results")
check_duplicates(sprint_quali, "Sprint Qualifying")
```

## Driver count per race

```python
def check_driver_counts(datasets, name):
    print(f"\n{name}")
    
    for i, df in enumerate(datasets):
        if "raceId" in df.columns:
            counts = df.groupby("raceId")["kaggle_driver_id"].nunique()
            print(f"Dataset {i} → min: {counts.min()}, max: {counts.max()}")
```

```python
check_driver_counts(main_race_results, "Main Race Results")
check_driver_counts(main_race_quali, "Main Race Qualifying")
check_driver_counts(sprint_results, "Sprint Race Results")
check_driver_counts(sprint_quali, "Sprint Qualifying")
```

## Check for missing critical values

```python
def check_missing_critical(datasets, name):
    print(f"\n{name}")
    
    for i, df in enumerate(datasets):
        print(f"\nDataset {i}")
        for col in ["raceId", "kaggle_driver_id", "Team", "Track"]:
            if col in df.columns:
                print(col, "→", df[col].isna().sum())
```

```python
check_missing_critical(main_race_results, "Main Race Results")
check_missing_critical(main_race_quali, "Main Race Qualifying")
check_missing_critical(sprint_results, "Sprint Race Results")
check_missing_critical(sprint_quali, "Sprint Qualifying")
```

## Quali Sanity check

```python
def check_quali_logic(datasets, name):
    print(f"\n{name}")
    
    for i, df in enumerate(datasets):
        if "participated_q2" in df.columns:
            invalid = ((df["participated_q3"] == 1) &
                       (df["participated_q2"] == 0)).sum()
            print(f"Dataset {i} invalid Q3 without Q2:", invalid)
```

```python
check_quali_logic(main_race_quali, "Main Race Qualifying")
check_quali_logic(sprint_quali, "Sprint Qualifying")
```
