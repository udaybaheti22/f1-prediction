## importing datasets

```python
circuits = pd.read_csv("filler-datasets/circuits.csv")
drivers = pd.read_csv("filler-datasets/drivers.csv")
results = pd.read_csv("filler-datasets/results.csv")
races = pd.read_csv("filler-datasets/races.csv")
```

### Removing unncessary colummns from Races

```python
races = races[['raceId', 'year','round','circuitId','name']]
```

```python
races.head()
```

### Removing unncessary colummns from Results

```python
results = results[['resultId','raceId','driverId','position','time','milliseconds','fastestLapTime','fastestLapSpeed']]
```

```python
drivers.head()
```

### Removing unncessary colummns from Races

```python
drivers = drivers[['driverId','driverRef','code','forename','surname']]
```

### Saving Temp Datasets

```python
races.to_csv('races_filtered.csv', index=False)
drivers.to_csv('drivers_filtered.csv', index=False)
results.to_csv('results_filtered.csv', index=False)
```

```python
races.info()
```

### Keeps rows where Year is between 2018 and 2025 (inclusive)

```python
# Keeps rows where Year is between 2018 and 2025 (inclusive)
races_filtered = races[races['year'].between(2018, 2025)]
```

```python
races_filtered.head()
```

#### Only keeping races between 2018 and 2025

```python
races_filtered.to_csv('races_filtered.csv', index=False)
results_filtered.to_csv('results_filtered.csv', index=False)
```

## Filtering Drivers

```python
drivers.info()
#we do not need all drivers
```

### Filtering drivers from 2018-2025

```python
values = results_filtered['driverId'].value_counts()
```

```python
values.info()
```

```python
unique_driver_ids = values.index.to_numpy()
```

```python
len(unique_driver_ids)
#40 unique drivers
```

### Mapping the driver's info from Drivers dataset

```python
# 1. Filter the dataset based on the NumPy array
# This keeps only rows where driverId is in your list
filtered_drivers = drivers[drivers['driverId'].isin(unique_driver_ids)]
# 2. Select only the specific columns you want
driver_dataset_ = filtered_drivers[['driverId', 'driverRef', 'code','forename','surname']]
```

```python
#final drivers dataset with drivers who participated from 2018
driver_dataset_.info()
```

```python
driver_dataset_.to_csv('drivers_filtered.csv', index=False)
```
